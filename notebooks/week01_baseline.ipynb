## 1. 載入資料與 Baseline

```python
from pathlib import Path
import sys

import pandas as pd
from sklearn.metrics import accuracy_score

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

sys.path.append(str(ROOT / "src"))

from rule_baseline import predict_batch

data_path = ROOT / "data" / "customer_intent_demo.csv"
df = pd.read_csv(data_path)
df.head()
```

## 2. 檢查資料筆數與類別分布

```python
print("rows:", len(df))
df["intent"].value_counts()
```

## 3. 執行 Rule-based Baseline

```python
df["prediction"] = predict_batch(df["message"].tolist())
df[["message", "intent", "prediction"]].head(10)
```

## 4. 計算 Accuracy

```python
accuracy = accuracy_score(df["intent"], df["prediction"])
print(f"Accuracy = {accuracy:.3f}")
```

## 5. Failure Cases

請從下面選一筆，填到 `evidence/week01_note.md`。

```python
failures = df[df["intent"] != df["prediction"]]
print("failure cases:", len(failures))
failures[["message", "intent", "prediction"]].head(5)
```
